```python
import os
import numpy as np
import pandas as pd

import rocks
```

```python
root = "/data/fink/fink-spins/"
data_path = os.path.join(root, "data", "ephemerides_atlas_ztf_2025.01.parquet")
sso_keys_path = os.path.join(data_path, "ssoname_keys.pkl")


# data_path = os.path.join(os.path.join(os.getcwd(), "data/ephemerides_atlas_ztf_2025.01.parquet"))
# sso_keys_path = os.path.join(os.path.join(os.getcwd(), "data/ssoname_keys.pkl"))
sso_keys = pd.read_pickle(sso_keys_path)
```

```python
info = pd.DataFrame(index=range(1000000), columns=["name"])
```

```python
i = 0
for k in list(sso_keys.keys()):
    
    df = pd.read_parquet(os.path.join(data_path, f"{k}"))
    for j, row in df.iterrows():

        info.loc[i,"name"] = row["name"]

        for f in [1,2,3,4]:
            cond = np.where( row.cfid == f )[0]

            info.loc[i,f"nobs_{f}"] = len( row.cjd[cond] )

            if len(row.cjd[cond]) > 0:
                info.loc[i,f"min_phase_{f}"] = np.min(row.Phase[cond])
                info.loc[i,f"max_phase_{f}"] = np.max(row.Phase[cond])
                info.loc[i,f"min_jd_{f}"] = np.min(row.cjd[cond])
                info.loc[i,f"max_jd_{f}"] = np.max(row.cjd[cond])
                
                            
        info.loc[i,f"nobs"] = len( row.cjd )
        info.loc[i,f"min_phase"] = np.min(row.Phase)
        info.loc[i,f"max_phase"] = np.max(row.Phase)
        info.loc[i,f"min_jd"] = np.min(row.cjd)
        info.loc[i,f"max_jd"] = np.max(row.cjd)


        i += 1

            


```
